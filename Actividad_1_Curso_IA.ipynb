1.Dime si la frase que te voy a dar es un palandrimoUn es una palabra que se lee igual al derecho y al revés (por ejemplo: "oso", "reconocer").

```python
frase=input("escribe una frase:")
derecho=frase.lower().replace(" ","")
reves= frase[::-1].lower().replace(" ","")
if derecho==reves:
    print("es un palíndromo")
else:
    print("No es un Palindromo")
```

2. Vocales
Dada una frase, cuentas cuantas vocales hay y dime cada tipo de ellas
Solicita al usuario una frase y cuenta cuántas vocales hay

```python
from itertools import count
frase2=input("escribre una frase:")
conteo_a = frase2.count('a')
conteo_e = frase2.count('e')
conteo_i = frase2.count('i')
conteo_o = frase2.count('o')
conteo_u = frase2.count('u')
conteo_a
conteo_e
conteo_i
conteo_o
conteo_u
print(f"El carácter 'A' aparece {conteo_a}")
print(f"El carácter 'E' aparece {conteo_e}")
print(f"El carácter 'I' aparece {conteo_i}")
print(f"El carácter 'O' aparece {conteo_o}")
print(f"El carácter 'U' aparece {conteo_u}")
```
