### ✅ 2. Normalización min-max
Normaliza el siguiente arreglo:
```python
arr = np.array([15, 22, 8, 30, 45])
```

````python
arr = np.array([15, 22, 8, 30, 45])
# Tu código aquí print(f"el maximo es {arr.max()} y minimo {arr.min()} ")
normalizados = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
normalizados
````

### ✅ 3. Multiplicación de matrices
Crea dos matrices de tamaño 3x3 con números aleatorios y realiza:
- Suma
- Producto elemento a elemento
- Producto matricial

````python
A = np.array([[1,8,5], [3,4,5],[3,4,2]])        # 2D
B = np.array([[4,2,9], [3, 4,6],[1,8,5]])        # 2D
print("Suma:", A+B)
print("Producto elemento a elemento:", A*B)
print("Producto matricial:", A@B)
````

### ✅ 4. Generar 100 números de una distribución normal
Luego grafica un histograma (usa matplotlib) y saca su media

````python
import matplotlib.pyplot as plt
# Tu código aquí

numeros_aleaotrios= np.random.normal(1, 100, size=100)
plt.hist(numeros_aleaotrios)
plt.show()
print("Media:", np.mean(numeros_aleaotrios), "Desviación estándar:", np.std(numeros_aleaotrios))
````

## 🐼 Ejercicios con pandas

### ✅ 1. Cargar un CSV desde URL
Usa este dataset:
```python
url = "https://people.sc.fsu.edu/~jburkardt/data/csv/hw_200.csv"
```
- Muestra las primeras 10 filas
- Calcula el promedio de peso y altura

````python
import pandas as pd

url = "https://people.sc.fsu.edu/~jburkardt/data/csv/hw_200.csv"
# Tu código aquí
df=pd.read_csv(url)
df.head(10)
````

### ✅ 2. Agrega una columna de clasificación por altura
Si la altura es mayor a 70, la persona es "alta", si no, es "baja"

````python
# Tu código aquí

df['altura'] = np.where(df[' Height(Inches)"'] > 70.0, "alta", "baja")
df['altura']
````

### ✅ 3. Filtrar datos
Filtra el DataFrame para mostrar solo las personas con peso mayor a 160.

````python
# Tu código aqui
df[df[' "Weight(Pounds)"'] > 160]

````

### ✅ 4. Agrupar y obtener estadísticas
Simula una columna de grupo (A, B o C) y:
- Agrupa por grupo
- Muestra el promedio de peso por grupo

````python
# Tu código aquí
df["grupo"] = ["A", "A", "B", "B", "C", "C"] * 33 + ["A", "B"]
promedios = df.groupby("grupo")[' "Weight(Pounds)"'].mean()
print(promedios)
````
