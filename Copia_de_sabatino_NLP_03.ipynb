## ¿Qué es el [NER](https://www.ibm.com/think/topics/named-entity-recognition)?

El reconocimiento de entidades nombradas (Named Entity Recognition), o extracción de entidades es una tarea del Procesamiento de Lenguaje Natural que identifica categorías predefinidas en los componentes de un texto.

Estas categorías son definidas por quienes entrenan el modelo, aunque algunos de los modelos preentrenados disponibles para su aplicación, son capaces de reconocer categorías como nombres propios, organizaciones, ubicaciones, expresiones de tiempo, cantidades, textos médics, moneda y porcentajes, entre otros.


### Técnicas de extracción de entidades
En general, hay varias técnicas para la aplicación del NER, aunque estas se pueden englobar en 3 tipos de técnicas diferentes:

* **Basadas en reglas:** al analizar la gramática de un lenguaje, podemos crear un conjunto de reglas que puedan ser usadas para identificar a estas entidades. Puntos en contra: son métodos tardados y no generalizan bien.
* **Basadas en aprendizaje automático (ML):** Hay que entrenar un modelo mediante un corpus etiquetado usando algoritmos como Árboles de decisión o SVMs, o algoritmos más complejos como redes neuronales o transformers. Puntos en contra: requieren grandes cantidades de datos etiquetados y pueden ser caros computacionalmente hablando.
* **Híbridos:** combina las dos anteriores aprovechando las fortalezas de ambas técnicas y compensando sus debilidades.

### El proceso
1. Recolección de datos: estos datos deben estar etiquetados. Las entidades en el texto se marcan y categorizan indicando sus tipos.
2. Preprocesamiento: Una vez obtenidos los datos, hay que preprocesarlos removiendo entidades no necesarias y tokenizando el texto.
3. Vectorización: durante este proceso, las características relevantes extraídas del texto preprocesado son extraídas como el POS, embeddings, información del contexto, entre otras para crear un nuevo vector que incluya todas estas características. La selección de características específicas dependerá de la tarea a resolvver.
4. Entrenamiento del modelo: Se entrena el modelo de ML o DL usando el conjunto de datos anotato y sus vectores de características.
5. Evaluación del modelo: Se evalúa el desemeño del modelo en datos nunca antes "vistos" por el mismo, para determinar mediante métricas como el F1-score que tan buen desempeño tiene el modelo para generalizar.
6. Fine-tunning: Dependiendo de la etapa de evaluación, se puede refinar el conocimiento del modelo para mejorar su desempeño al ajustar hiperparámetros, modificar las características o añadir técnicas nuevas.
7. Inferencia: En esta etapa el modelo está listo para realizar inferencia sobre datos totalmente nuevos. En este proceso, se toma un texto nuevo, se preprocesa, se extraen características y se predice.

### 2. Importamos las librerías que vamos a utilizar

```python
import os

os.environ["KERAS_BACKEND"] = "tensorflow"

import keras
from keras import ops
import numpy as np
import tensorflow as tf
from keras import layers
from datasets import load_dataset
from collections import Counter
from conlleval import evaluate
```

### 3. Usemos keras para crear una capa de transformer

```python
class TransformerBlock(layers.Layer):
  def __init__(self, embed_dim, num_heads, ff_dim, rate=0.1):
    super().__init__()
    self.att = keras.layers.MultiHeadAttention(
      num_heads=num_heads, key_dim=embed_dim
    )
    self.ffn = keras.Sequential(
      [
        keras.layers.Dense(ff_dim, activation="relu"),
        keras.layers.Dense(embed_dim),
      ]
    )
    self.layernorm1 = keras.layers.LayerNormalization(epsilon=1e-6)
    self.layernorm2 = keras.layers.LayerNormalization(epsilon=1e-6)
    self.dropout1 = keras.layers.Dropout(rate)
    self.dropout2 = keras.layers.Dropout(rate)

  def call(self, inputs, training=False):
    attn_output = self.att(inputs, inputs)
    attn_output = self.dropout1(attn_output, training=training)
    out1 = self.layernorm1(inputs + attn_output)
    ffn_output = self.ffn(out1)
    ffn_output = self.dropout2(ffn_output, training=training)

    return self.layernorm2(out1 + ffn_output)
```

### 4. Definamos una capa de tokens y de embeddings posicionales

```python
class TokenAndPositionEmbedding(layers.Layer):
  def __init__(self, maxlen, vocab_size, embed_dim):
    super().__init__()
    self.token_emb = keras.layers.Embedding(
      input_dim=vocab_size, output_dim=embed_dim
    )
    self.pos_emb = keras.layers.Embedding(input_dim=maxlen, output_dim=embed_dim)

  def call(self, inputs):
    maxlen = ops.shape(inputs)[-1]
    positions = ops.arange(start=0, stop=maxlen, step=1)
    position_embeddings = self.pos_emb(positions)
    token_embeddings = self.token_emb(inputs)
    return token_embeddings + position_embeddings
```

### 5. Construimos el modelo NER

```python
class NERModel(keras.Model):
  def __init__(
    self, num_tags, vocab_size, maxlen=128, embed_dim=32, num_heads=2, ff_dim=32
  ):
    super().__init__()
    self.embedding_layer = TokenAndPositionEmbedding(maxlen, vocab_size, embed_dim)
    self.transformer_block = TransformerBlock(embed_dim, num_heads, ff_dim)
    self.dropout1 = layers.Dropout(0.1)
    self.ff = layers.Dense(ff_dim, activation="relu")
    self.dropout2 = layers.Dropout(0.1)
    self.ff_final = layers.Dense(num_tags, activation="softmax")

  def call(self, inputs, training=False):
    x = self.embedding_layer(inputs)
    x = self.transformer_block(x)
    x = self.dropout1(x, training=training)
    x = self.ff(x)
    x = self.dropout2(x, training=training)
    x = self.ff_final(x)
    return x
```

#### Exportamos a un archivo txt para fácil lectura

```python
def export_to_file(export_file_path, data):
  with open(export_file_path, "w") as f:
    for record in data:
      ner_tags = record["ner_tags"]
      tokens = record["tokens"]
      if len(tokens) > 0:
        f.write(
          str(len(tokens))
          + "\t"
          + "\t".join(tokens)
          + "\t"
          + "\t".join(map(str, ner_tags))
          + "\n"
        )


os.mkdir("data")
export_to_file("/content/data/conll_train.txt", conll_data["train"])
export_to_file("/content/data/conll_val.txt", conll_data["validation"])
```

### 7. Definamos nuestras categorías de las entidades

```python
def make_tag_lookup_table():
  iob_labels = ["B", "I"]
  ner_labels = ["PER", "ORG", "LOC", "MISC"]
  all_labels = [(label1, label2) for label2 in ner_labels for label1 in iob_labels]
  all_labels = ["-".join([a, b]) for a, b in all_labels]
  all_labels = ["[PAD]", "O"] + all_labels
  return dict(zip(range(0, len(all_labels) + 1), all_labels))


mapping = make_tag_lookup_table()
print(mapping)
```

### 8. Creamos una lista de todos los tokens en el conjunto de entrenamiento que será usada para crear el vocabulario

```python
all_tokens = sum(conll_data["train"]["tokens"], [])
all_tokens_array = np.array(list(map(str.lower, all_tokens)))

counter = Counter(all_tokens_array)
print(len(counter))

num_tags = len(mapping)
vocab_size = 20000

'''
Nota: tomamos las vocab_size - 2 palabras más frecuentes del conjunto de
entrenamiento ya que 'StringLookup' usa 2 tokens adicionales para
tokens desconocidos y el de enmascaramiento.
'''
vocabulary = [token for token, count in counter.most_common(vocab_size - 2)]

# StringLook convierte los tokens a sus IDs
lookup_layer = keras.layers.StringLookup(vocabulary=vocabulary)
```

### 9. Importamos los conjuntos de entrenamiento y validación

```python
train_data = tf.data.TextLineDataset("/content/data/conll_train.txt")
val_data = tf.data.TextLineDataset("/content/data/conll_val.txt")
```

```python
print(list(train_data.take(1).as_numpy_iterator()))
```

```python
def map_record_to_training_data(record):
  record = tf.strings.split(record, sep="\t")
  length = tf.strings.to_number(record[0], out_type=tf.int32)
  tokens = record[1 : length + 1]
  tags = record[length + 1 :]
  tags = tf.strings.to_number(tags, out_type=tf.int64)
  tags += 1
  return tokens, tags


def lowercase_and_convert_to_ids(tokens):
  tokens = tf.strings.lower(tokens)
  return lookup_layer(tokens)


'''
Usamos el padded_batch ya que cada entrada en el corpus tiene una longitud
variable.
'''
batch_size = 32
train_dataset = (
  train_data.map(map_record_to_training_data)
  .map(lambda x, y: (lowercase_and_convert_to_ids(x), y))
  .padded_batch(batch_size)
)
val_dataset = (
  val_data.map(map_record_to_training_data)
  .map(lambda x, y: (lowercase_and_convert_to_ids(x), y))
  .padded_batch(batch_size)
)

ner_model = NERModel(num_tags, vocab_size, embed_dim=32, num_heads=4, ff_dim=64)
```

### 10. Usamos una función objetivo personalizada para el modelo NER

```python
class CustomNonPaddingTokenLoss(keras.losses.Loss):
  def __init__(self, name="custom_ner_loss"):
    super().__init__(name=name)

  def call(self, y_true, y_pred):
    loss_fn = keras.losses.SparseCategoricalCrossentropy(
      from_logits=False, reduction=None
    )
    loss = loss_fn(y_true, y_pred)
    mask = ops.cast((y_true > 0), dtype="float32")
    loss = loss * mask
    return ops.sum(loss) / ops.sum(mask)


loss = CustomNonPaddingTokenLoss()
```

### 10. Compilamos y entrenamos el modelo

```python
tf.config.run_functions_eagerly(True)
ner_model.compile(optimizer="adam", loss=loss)
ner_model.fit(train_dataset, epochs=10)


def tokenize_and_convert_to_ids(text):
    tokens = text.split()
    return lowercase_and_convert_to_ids(tokens)
```

```python
def calculate_metrics(dataset):
    all_true_tag_ids, all_predicted_tag_ids = [], []

    for x, y in dataset:
        output = ner_model.predict(x, verbose=0)
        predictions = ops.argmax(output, axis=-1)
        predictions = ops.reshape(predictions, [-1])

        true_tag_ids = ops.reshape(y, [-1])

        mask = (true_tag_ids > 0) & (predictions > 0)
        true_tag_ids = true_tag_ids[mask]
        predicted_tag_ids = predictions[mask]

        all_true_tag_ids.append(true_tag_ids)
        all_predicted_tag_ids.append(predicted_tag_ids)

    all_true_tag_ids = np.concatenate(all_true_tag_ids)
    all_predicted_tag_ids = np.concatenate(all_predicted_tag_ids)

    predicted_tags = [mapping[tag] for tag in all_predicted_tag_ids]
    real_tags = [mapping[tag] for tag in all_true_tag_ids]

    evaluate(real_tags, predicted_tags)


calculate_metrics(val_dataset)
```
