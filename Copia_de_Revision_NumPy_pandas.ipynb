## 🔢 NumPy: funciones esenciales

```python
import numpy as np

# Crear arreglos
arr1 = np.array([1, 2, 3])
arr2 = np.array([[1, 2], [3, 4]])
zeros = np.zeros((2, 2))
ones = np.ones((2, 2))
identity = np.eye(3)

print("Arreglo:", arr1)
print("Matriz 2x2:", arr2)
print("Zeros:", zeros)
```

```python
# Propiedades
print("Forma:", arr2.shape)
print("Dimensiones:", arr2.ndim)
print("Tipo de dato:", arr2.dtype)
print("Tamaño:", arr2.size)
```

```python
# Operaciones
print("Suma:", arr1 + 2)
print("Producto escalar:", arr1 * 2)
print("Raíz cuadrada:", np.sqrt(arr1))
print("Suma total:", np.sum(arr1))
```

```python
# Indexación
print("Primer elemento:", arr1[0])
print("Slicing:", arr1[1:])
print("Transpuesta:", arr2.T)
```

## 🐼 pandas: funciones esenciales

```python
import pandas as pd

# Crear un DataFrame
data = {
    'nombre': ['Ana', 'Luis', 'Marta'],
    'edad': [23, 25, 22],
    'ciudad': ['CDMX', 'Puebla', 'GDL']
}
df = pd.DataFrame(data)
print(df)
```

```python
# Exploración básica
print(df.head())
print(df.info())
print(df.describe())
```

```python
# Selección y filtrado
print(df['edad'])
print(df[df['edad'] > 23])
```

```python
# Modificación de columnas
df['mayor_edad'] = df['edad'] >= 18
print(df)
```

```python
# Agrupación y resumen
print(df.groupby('ciudad')['edad'].mean())
```
