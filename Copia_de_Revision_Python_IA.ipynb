## 1. Estructuras de datos básicas

**Listas**
Estructuras de datos que permiten almacenar una colección ordenada y modificable
se pueden guardar cualquier tipo de datos: números, cadenas, booleanos e incluso otras listas
son mutables, es decir, se pueden modificar una vez creadas (CRUD)


```python

#######################################

# Crear una lista
mi_lista = [10, 20, 30, 40]

# Acceder a elementos
print(mi_lista[0])    # 10
print(mi_lista[-1])   # 40 (último)

# Modificar un elemento
mi_lista[1] = 25      # Cambia 20 por 25

# Agregar elementos
mi_lista.append(50)

# Eliminar elementos
mi_lista.remove(30)   # Elimina el primer 30 que encuentre

# Longitud
print(len(mi_lista))  # 4
```

**Diccionarios**
Son estructuras de datos que almacenan pares de clave y valor, la principal diferencia con las listas es que accedes a los elementos por clave y no por posición
Cada entrada tiene una clave(key) y un valor (value)
Las claves deben ser únicas e inmutables (ex. string o números)
son mutables
Crear un diccionario


```python

#################################

persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "CDMX"
}

# Acceder a valores
print(persona["nombre"])   # Ana

# Agregar un nuevo par
persona["profesion"] = "Ingeniera"

# Modificar un valor
persona["edad"] = 31

# Eliminar un par
del persona["ciudad"]

# Recorrer claves y valores
for clave, valor in persona.items():
    print(clave, "=>", valor)
```

## 2. Funciones en Python

**Funciones**: Una función es un bloque de código reutilizable que realiza una tarea específica, sirve para organizar, modularizar y reutilizar cóodigo  sin necesidad de repetirlo

```python
def cuadrado(x):
    return x**2

print(cuadrado(5))

# Función lambda
cuadrado_lambda = lambda x: x**2
print(cuadrado_lambda(6))
```

## 3. NumPy para operaciones numéricas

```python
import numpy as np #vamos a importar numpy con su alias

```

```python
#Crear arreglos
a = np.array([1, 2, 3])               # 1D
b = np.array([[1, 2], [3, 4]])        # 2D
c = np.zeros((2, 3))                  # Matriz de ceros
d = np.ones((2, 2))                   # Matriz de unos
e = np.eye(3)                         # Matriz identidad
f = np.arange(0, 10, 2)               # [0 2 4 6 8]
g = np.linspace(0, 1, 5)              # [0. 0.25 0.5 0.75 1.]
```

```python
#Propiedades de los array
a.shape        # Tamaño del array
a.ndim         # Número de dimensiones
a.dtype        # Tipo de dato
a.size         # Número total de elementos
```

```python
#Operaciones elementales

a = np.array([1, 2, 3])               # 1D
z = np.array([5, 2, 3])               # 1D


a + a          # Suma elemento a elemento
a * 2          # Multiplicación escalar
np.sqrt(a)     # Raíz cuadrada
np.exp(a)      # Exponencial
np.dot(a, z)   # Producto matricial
```

```python
#Estadisticos
np.mean(a)     # Promedio
np.std(a)      # Desviación estándar
np.min(a)      # Mínimo
np.max(a)      # Máximo
np.sum(a)      # Suma total
```

```python
#Indexación y slicing
a[0]           # Primer elemento
a[-1]          # Último elemento
a[1:4]         # Slicing
b[0, 1]        # Elemento fila 0, columna 1
b[:, 1]        # Toda la columna 1
```

```python
#Reshape y transposición
a.reshape((3, 1))    # Cambiar forma
a.T                 # Transpuesta
```

```python
#Concatenar
np.concatenate([a, z], axis=0)   # Unir arreglos
```

```python
#Comparación de máscaras booleanas
a > 2                     # Devuelve booleanos
a[a > 2]                 # Filtrado
np.where(a > 2, 1, 0)     # Condicional
```

```python
#Copias y vistas
b = a.copy()             # Copia real (no cambia si cambia a)
```

## 7. Primer modelo de Machine Learning

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42)

clf = RandomForestClassifier()
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)

print("Precisión:", accuracy_score(y_test, y_pred))
```
