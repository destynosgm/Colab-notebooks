```python

x= student["study_hours_per_day"]
y=student["exam_score"]

beta0=28
beta1=11
yp=beta0+beta1*x


plt.scatter(x,y)
plt.plot(x,yp,"r--")
```

```python
titanic.head()
titanic_females=titanic[titanic["Sex"]=="female"]
titanic_females[["Fare","Survived"]].sample(15)
```

```python
import seaborn as sns
sns.regplot(x="Fare",y="Survived",data=titanic_females,logistic=True)
#titanic_females.plot.scatter("Fare","Survived")

```

```python
from numpy import exp
x=titanic_females["Fare"]
y=titanic_females["Survived"]
BETA0=0.6
beta1=0.01

yp=exp(BETA0+beta1*x)/(1+exp(BETA0+beta1*x))

plt.scatter(x,y)
plt.plot(x,yp,"r*")
```

```python
pd.DataFrame({
    "x":x,
    "y":y,
    "yp":yp,
    "e":y-yp,
    "e2":(y-yp)**2
})["e2"].sum()
```

```python
from sklearn.linear_model import LogisticRegression

reg= LogisticRegression()
z=x.values.reshape(-1,1)
reg.fit(z,y)
pd.DataFrame({
    "beta0":[reg.intercept_[0]],
    "beta1":[reg.coef_[0][0]],
    "score":[reg.score(z,y)],
})
```

```python
BETA0=reg.intercept_
beta1=reg.coef_[0]

yp=exp(BETA0+beta1*x)/(1+exp(BETA0+beta1*x))
pd.DataFrame({
    "x":x,
    "y":y,
    "yp":yp,
    "e":y-yp,
    "e2":(y-yp)**2
})["e2"].sum()
```
